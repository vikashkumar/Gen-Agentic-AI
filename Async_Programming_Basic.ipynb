```python
import time

def fetch_student(student_id):
    print(f"Fetching student {student_id}...")

    # Simulates waiting for an API or database
    time.sleep(20)

    print(f"Student {student_id} received")
    return {"student_id": student_id}


start_time = time.perf_counter()

fetch_student(101)
fetch_student(102)
fetch_student(103)

end_time = time.perf_counter()

print(f"\nTotal time: {end_time - start_time:.2f} seconds")
```

```python
import asyncio
import time


async def fetch_student(student_id):
    print(f"Started fetching student {student_id}")

    # Simulate waiting for an API or database
    await asyncio.sleep(20)

    print(f"Finished fetching student {student_id}")
    return {"student_id": student_id}


async def main():
    start_time = time.perf_counter()

    results = await asyncio.gather(
        fetch_student(101),
        fetch_student(102),
        fetch_student(103)
    )

    end_time = time.perf_counter()

    print("\nResults:", results)
    print(f"Total time: {end_time - start_time:.2f} seconds")


# Use this in Google Colab or Jupyter Notebook
await main()
```
