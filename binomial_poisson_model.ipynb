```python
import math

binomial_model = {
    "type": "binomial",
    "n": 10,
    "p": 0.2
}

#print(binomial_model["n"])

poisson_model = {
    "type": "poisson",
    "rate" : 2
}


def pmf(model, k):
    if model["type"] == "binomial":
      n = model["n"]
      p = model["p"]
      return math.comb(n, k) * p**k * (1-p)**(n-k)

    elif model["type"] == "poisson":
      rate = model["rate"]

      return (
          math.exp(-rate)     #e raise to - lambda (rate)
          * rate ** k        # lambda raise to power of k
          / math.factorial(k) # k factorial
      )

  #straight line formula
  #(math.exp(-rate) * rate ** k) / math.factorial(k)

def mean(model):
  if model["type"] == "binomial":
    n = model["n"]
    p = model["p"]
    return n * p

  elif model["type"] == "poisson":
    rate = model["rate"]
    return rate

def variance(model):
  if model["type"] == "binomial":
    n = model["n"]
    p = model["p"]
    return n * p * (1-p)
  elif model["type"] == "poisson":
    rate = model["rate"]
    return rate

models = [binomial_model, poisson_model]

for model in models:
  print(
      model["type"],
      "Probability:", pmf(model, 4),
      "Mean:", mean(model),
      "Variance:",variance(model)
  )

```

```python
class DiscreteDistribution:

  def pmf(self, k):
    raise NotImplementedError

  def mean(self):
    raise NotImplementedError

  def variance(self):
    raise NotImplementedError
```

```python
class Binomial(DiscreteDistribution):
  def __init__(self, n, p):
    if not isinstance(n, int) or n <=0:
      raise ValueError(
          "n must be a positive integer"
      )

    if not (0 <= p <= 1):
      raise ValueError(
          "p must be between 0 and 1"
      )

    self.n = n
    self.p = p

  def pmf(self, k):
    if not isinstance(k, int) or k < 0:
      raise ValueError(
          "k must be a positive integer"
      )

    return math.comb(self.n, k) * self.p**k * (1-self.p)**(self.n-k)

  def mean(self):
    return self.n * self.p

  def variance(self):
    return self.n * self.p * (1-self.p)




```

```python
class Poisson(DiscreteDistribution):
  def __init__(self, rate):
    if rate<=0:
      raise ValueError(
          "rate must be greater than zero."
      )

    self.rate = rate

  def pmf(self,k):
    if not isinstance(k, int):
      raise TypeError("k must be an integer")

    return (
        math.exp(-self.rate)     #e raise to - lambda (rate)
        * self.rate ** k        # lambda raise to power of k
        / math.factorial(k)     # k factorial
    )

  def mean(self):
    return self.rate

  def variance(self):
    return self.rate
```

```python
def print_report(distribution, k):
  name = type(distribution).__name__

  print(f"\n{name}")
  print(f"Probability: {distribution.pmf(k)}")
  print(f"Mean: {distribution.mean()}")
  print(f"Variance: {distribution.variance()}")
```

```python
models =[
    Binomial(n=10, p = 0.2),
    Poisson(rate = 2)
]

for model in models:
  #print(model)
  print_report(model, k=2)
```
